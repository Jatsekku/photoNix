This example uses *gdsfactory* and *tidy3d* to calculates modes propagating through waveguide.
Based on: https://gdsfactory.github.io/gplugins/notebooks/mode_solver_fdfd/

```python
import gdsfactory as gf
import gplugins.tidy3d as gt
import matplotlib.pyplot as plt
import numpy as np
from gdsfactory.gpdk import get_generic_pdk
```

```python
gf.config.rich_output()
PDK = get_generic_pdk()
PDK.activate()

strip = gt.modes.Waveguide(
    # Wavelength of interest (1.55um)
    wavelength=1.55,
    # Waveguide width (500nm)
    core_width=0.5,
    # Waveguide height (220nm)
    core_thickness=0.22,
    # Core material
    core_material="si",
    # Clad material
    clad_material="sio2",
    # Number of modes to compute
    num_modes=2,
)

strip.plot_index()
strip.plot_grid()
strip.plot_field(field_name="Ex", mode_index=0)

```
